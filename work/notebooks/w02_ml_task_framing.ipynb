## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Ranking / scoring:** prioritize already-visible pages for editorial review. One row represents one page in the starter snapshot. The output is a ranked queue; editors review the evidence before deciding whether to refresh.

**Target proxy:** the starter label is `trend_direction == "down"`, derived from observed recent changes. It is a retrospective proxy for decline, not evidence that a refresh will work. A future-facing model must define a decision date and construct its target from a later, non-overlapping window.

**Metric:** Precision@50 on held-out clients measures the proportion of the top 50 flagged pages that actually decline. Compare to the same fixed-rule baseline and report uncertainty and per-client variation. A useful model should improve on that baseline on the same held-out rows; no universal threshold is assumed.

**Why consider ML?** Page age, exposure and position may interact, and learned scores can combine signals that fixed thresholds miss. This is a hypothesis, not a guaranteed advantage: use a simple baseline, exclude outcome-derived features, validate by client and time, and prefer the simpler rule if the improvement does not hold. False positives cost editorial time; missed declining pages delay review.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd
url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)
pages = df.loc[df.impressions_90d > 0, ["impressions_90d", "content_age_days", "days_since_last_update", "avg_position", "trend_direction"]].copy()
pages["declining_proxy"] = pages.pop("trend_direction").str.lower().eq("down").astype(int)
print("One row = one page in the starter snapshot; pages:", len(pages))
display(pages.head())
```
